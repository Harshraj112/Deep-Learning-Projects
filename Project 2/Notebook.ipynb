```python
#import libraries.
import keras
import tensorflow as tf
```

```python
mnist=tf.keras.datasets.mnist
#split data
(x_train, y_train), (x_test, y_test) = mnist.load_data()
```

```python
x_train, x_test = x_train / 255.0, x_test / 255.0
```

```python
digit = tf.keras.models.Sequential([tf.keras.layers.Flatten(input_shape=(28, 28)),tf.keras.layers.Dense(10, activation=tf.nn.softmax)])
#compile model
digit.compile(optimizer='adam',loss='sparse_categorical_crossentropy',metrics=['accuracy'])
```

```python
digit.fit(x_train, y_train, epochs=3)
digit.evaluate(x_test, y_test)
```
